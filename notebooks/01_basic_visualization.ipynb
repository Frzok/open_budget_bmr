# Анализ финансовых данных

```

# Анализ финансовых данных Балтийского городского совета

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Загрузка данных
df = pd.read_csv('../data/raw/sample_financial_data.csv')

# Группировка по году и категории
pivot_df = df.pivot(index='Категория', columns='Год', values=['Доходы (тыс. €)', 'Расходы (тыс. €)'])

# Визуализация доходов и расходов по категориям
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Категория', y='Доходы (тыс. €)', hue='Год')
plt.title('Доходы по категориям и годам')
plt.ylabel('Доходы (тыс. €)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Категория', y='Расходы (тыс. €)', hue='Год')
plt.title('Расходы по категориям и годам')
plt.ylabel('Расходы (тыс. €)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
